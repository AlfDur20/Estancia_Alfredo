```python
import yfinance as yf
import pandas as pd

# Celda 0: descargar y graficar el ticker "SYP" entre 2022-01-01 y 2024-01-01
import matplotlib.pyplot as plt

ticker = "SPY"
START_DATE='2022-01-01'
END_DATE='2025-11-01'
AN_DATE='2024-01-01'

data = yf.download(ticker, start=START_DATE, end=END_DATE, progress=False)

if data.empty:
    print(f"No se encontraron datos para el ticker '{ticker}'. Verifica el símbolo.")
else:
    plt.figure(figsize=(12,6))
    plt.plot(data.index, data['Close'], label=f"{ticker} - Cierre")
    plt.title(f"{ticker}: Precio de cierre ({START_DATE} a {END_DATE})")
    # Línea vertical en AN_DATE para separar periodos
    an_dt = pd.to_datetime(AN_DATE)
    plt.axvline(x=an_dt, color='red', linestyle='--', linewidth=1.5, label=f"Separador: {AN_DATE}")
    plt.xlabel("Fecha")
    plt.ylabel("Precio")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```
